```python
# ==========================================
# CAB BOOKING & FARE MANAGEMENT SYSTEM
# ==========================================

import random # Import the random module to generate random OTP numbers

rides = [] # Create an empty list to store all booked rides

# ==========================================
# FUNCTION 1: DISPLAY MENU
# ==========================================
def display_menu():
    print("\n" + "=" * 45) 
    print("          CAB BOOKING SYSTEM") 
    print("=" * 45)   
    print("1. Book Ride")
    print("2. View Ride History")
    print("3. Search Ride")
    print("4. Revenue Report")
    print("5. Find Longest Ride")
    print("6. Find Highest Fare")
    print("7. Daily Summary Report")
    print("8. Exit")
    print("=" * 45)   # Print another separator line

# ==========================================
# INPUT VALIDATION
# ==========================================
def get_valid_name(prompt):
    while True:
        name = input(prompt).strip()

        if name and any(char.isalpha() for char in name):
            return name

        print("Invalid name! Please enter a valid name.")


def get_valid_location(prompt):
    while True:
        location = input(prompt).strip()

        if location:
            return location

        print("Location cannot be empty!")


def get_valid_distance():
    while True:
        try:
            distance = float(input("Enter Distance (KM): "))

            if distance > 0:
                return distance

            print("Distance must be greater than zero.")

        except ValueError:
            print("Invalid input! Enter a numeric distance.")

# ==========================================
# FUNCTION 2: RANDOM OTP VERIFICATION
# ==========================================
def verify_otp():
    otp = random.randint(100000, 999999)

    print("\n========== OTP VERIFICATION ==========")
    print("Your Demo OTP is:", otp)

    attempts = 3

    while attempts > 0:
        entered_otp = input(
            "Enter OTP to Book a Ride: "
        ).strip()

        if entered_otp == str(otp):
            print("OTP Verified Successfully!")
            return True

        attempts -= 1
        print("Invalid OTP!")

        if attempts > 0:
            print("Attempts remaining:", attempts)

    print("OTP verification failed. Booking cancelled.")
    return False

# ==========================================
# FUNCTION 3: CALCULATE FARE
# ==========================================
def calculate_fare(distance):
    base_fare = 50

    if distance <= 5:
        rate_per_km = 15
    elif distance <= 15:
        rate_per_km = 12
    else:
        rate_per_km = 10

    total_fare = base_fare + (distance * rate_per_km)

    return round(total_fare, 2)

# ==========================================
# FUNCTION 4: BOOK A RIDE
# ==========================================
def book_ride():
    print("\n========== BOOK A RIDE ==========")

    # Verify OTP before collecting booking details
    if not verify_otp():
        return

    customer_name = get_valid_name("Enter Customer Name: ")
    pickup = get_valid_location("Enter Pickup Location: ")
    destination = get_valid_location("Enter Destination: ")
    distance = get_valid_distance()

    fare = calculate_fare(distance)

    ride = {
        "customer_name": customer_name,
        "pickup": pickup,
        "destination": destination,
        "distance": distance,
        "fare": fare
    }

    rides.append(ride)

    print("\nRide booked successfully!")
    print("-" * 35)
    print("Customer Name :", customer_name)
    print("Pickup        :", pickup)
    print("Destination   :", destination)
    print(f"Distance      : {distance:g} KM")
    print(f"Total Fare    : ₹{fare:.2f}")
    print("-" * 35)

# ==========================================
# FUNCTION 5: VIEW RIDE HISTORY
# ==========================================
def view_rides():
    print("\n========== RIDE HISTORY ==========")

    if not rides:
        print("No rides have been booked yet.")
        return

    print(
        f"{'Customer':<16}"
        f"{'Pickup':<16}"
        f"{'Destination':<16}"
        f"{'Distance':<12}"
        f"{'Fare':>10}"
    )
    print("-" * 70)

    for ride in rides:
        print(
            f"{ride['customer_name'][:15]:<16}"
            f"{ride['pickup'][:15]:<16}"
            f"{ride['destination'][:15]:<16}"
            f"{ride['distance']:>7g} KM  "
            f"₹{ride['fare']:>8.2f}"
        )

    print("-" * 70)
    print("Total Rides:", len(rides))

# ==========================================
# FUNCTION 6: SEARCH RIDE BY CUSTOMER
# ==========================================
def search_ride():
    print("\n========== SEARCH RIDE ==========")

    if not rides:
        print("No rides available to search.")
        return

    search_name = get_valid_name(
        "Enter Customer Name to Search: "
    )

    found_rides = [
        ride for ride in rides
        if ride["customer_name"].lower() == search_name.lower()
    ]

    if not found_rides:
        print("No ride found for this customer.")
        return

    for index, ride in enumerate(found_rides, start=1):
        print(f"\nRide {index}")
        print("-" * 35)
        print("Customer Name :", ride["customer_name"])
        print("Pickup        :", ride["pickup"])
        print("Destination   :", ride["destination"])
        print(f"Distance      : {ride['distance']:g} KM")
        print(f"Fare          : ₹{ride['fare']:.2f}")

# ==========================================
# FUNCTION 7: REVENUE REPORT
# ==========================================
def revenue_report():
    print("\n========== REVENUE REPORT ==========")

    if not rides:
        print("No rides available to generate a report.")
        return

    total_rides = len(rides)
    total_revenue = sum(ride["fare"] for ride in rides)
    average_fare = total_revenue / total_rides

    print("Total Rides Completed :", total_rides)
    print(f"Total Revenue         : ₹{total_revenue:.2f}")
    print(f"Average Fare Amount   : ₹{average_fare:.2f}")

# ==========================================
# BONUS 1: FIND LONGEST RIDE
# ==========================================
def find_longest_ride():
    print("\n========== LONGEST RIDE ==========")

    if not rides:
        print("No rides available.")
        return

    longest = max(rides, key=lambda ride: ride["distance"])

    print("Customer Name :", longest["customer_name"])
    print("Pickup        :", longest["pickup"])
    print("Destination   :", longest["destination"])
    print(f"Distance      : {longest['distance']:g} KM")

# ==========================================
# BONUS 2: FIND HIGHEST FARE
# ==========================================
def find_highest_fare():
    print("\n========== HIGHEST FARE ==========")

    if not rides:
        print("No rides available.")
        return

    highest = max(rides, key=lambda ride: ride["fare"])

    print("Customer Name :", highest["customer_name"])
    print("Pickup        :", highest["pickup"])
    print("Destination   :", highest["destination"])
    print(f"Highest Fare  : ₹{highest['fare']:.2f}")

# ==========================================
# BONUS 3: DAILY SUMMARY REPORT
# ==========================================
def daily_summary_report():
    print("\n========== DAILY SUMMARY ==========")

    if not rides:
        print("No rides available.")
        return

    total_rides = len(rides)
    total_revenue = sum(ride["fare"] for ride in rides)

    average_distance = sum(
        ride["distance"] for ride in rides
    ) / total_rides

    highest = max(rides, key=lambda ride: ride["fare"])

    print("Total Rides       :", total_rides)
    print(f"Total Revenue     : ₹{total_revenue:.2f}")
    print(f"Average Distance  : {average_distance:.2f} KM")
    print(f"Highest Fare      : ₹{highest['fare']:.2f}")
    print("Highest Fare Customer:", highest["customer_name"])

# ==========================================
# MAIN PROGRAM
# ==========================================
def main():
    while True:
        display_menu()

        choice = input("Enter Your Choice: ").strip()

        if choice == "1":
            book_ride()

        elif choice == "2":
            view_rides()

        elif choice == "3":
            search_ride()

        elif choice == "4":
            revenue_report()

        elif choice == "5":
            find_longest_ride()

        elif choice == "6":
            find_highest_fare()

        elif choice == "7":
            daily_summary_report()

        elif choice == "8":
            print("\nThank you for using Cab Booking System!")
            print("Have a safe journey!")
            break

        else:
            print("Invalid choice! Please enter a number from 1 to 8.")

# ==========================================
# START APPLICATION
# ==========================================
if __name__ == "__main__":
    main()
```
